```python
from langchain_openai import OpenAIEmbeddings
from langchain_community.vectorstores import Chroma
```

```python
from langchain_core.documents import Document
#create lanchain documets for IPL Players
doc1=Document(
    page_content="Virat Kohli is one of the most successful and consistent batsment in IPL history. known for his aggressive batting style",
    metadata={"team":"Royal challengers Bangalore"}
)
doc2=Document(
    page_content="Rohit Sharma is one of the most successful captain in IPL history. leading Mumbai Indians to five titles",
    metadata={"team":"Mumbai Indians"}
)
doc3=Document(
    page_content="MS Dhoni, famously known as captain cool, has led Chennai Super kings to multiple IPL titles. ",
    metadata={"team":"Chennai Super kings"}
)
doc4=Document(
    page_content="Jasprit Bumrah in considered one of the best fast bowlers in T20 cricket playing for Mumbai Indians",
    metadata={"team":"Mumbai Indians"}
)
```

```python
docs=[doc1,doc2,doc3,doc4]
```

```python
import os
os.environ["OPENAI_API_KEY"] = "api key"# replace this key while you are running
Vector_store=Chroma(
    embedding_function=OpenAIEmbeddings(),
    persist_directory='My_chroma_db',
    collection_name='sample'
)
#this will create a folder My_chroma_db and it will also create chroma.sqlite3 database for our query search
```

```python
#add documents
#Vector_store.add_documents(docs)
#this will create a id for each docs which we are passing and this we later use it to modify the exiting data
ids = ["doc1", "doc2", "doc3", "doc4"]

Vector_store.add_documents(
    documents=docs,
    ids=ids
)
```

```python
#use this to delete the vector which was created with repetative ids,
#Vector_store.delete_collection()
```

```python
#view documents
Vector_store.get(
    include=["documents", "metadatas", "embeddings"]
)
```

```python
#search documents
Vector_store.similarity_search(
    query="who among these are bowler?", #input query
    k=1 #how much similar object we want in output
)
```

```python
#search similarity with score
#search documents
Vector_store.similarity_search_with_score(
    query="who among these are bowler?", #input query
    k=2
)
```

```python
# Filter documents based on team
Vector_store.similarity_search(
    query="", # An empty query can be used if you just want to filter by metadata
    k=4, # Return up to 4 results (or adjust as needed)
    filter={"team": "Mumbai Indians"}
)
```

```python
#update documents
updated_doc1=Document(
    page_content="Virat Kohli, the former captain of Royal Challengers Bangalore (RCB), is renowned for his aggressive batting style",
    metadata={"team":"Royal challengers Bangalore"}
)
Vector_store.update_document(
    document=updated_doc1,
    document_id="doc1"
)
```

```python
#view documents
Vector_store.get(include=["documents", "metadatas", "embeddings"])
```

```python
#delete document
Vector_store.delete(ids=["doc2"])
#view documents after deleting the doc2
Vector_store.get(include=["documents", "metadatas", "embeddings"])
```
