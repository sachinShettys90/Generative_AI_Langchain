```python
from langchain_community.vectorstores import Chroma
from langchain_openai import OpenAIEmbeddings
from langchain_core.documents import Document
```

```python
documents=[
    Document(page_content="langchain is a developer friendly framework"),
    Document(page_content="Chroma is a vector data base for LLM"),
    Document(page_content="AI is an artificial intelligence ")
]


embedding_model = OpenAIEmbeddings(api_key="api key")
```

```python
vectorStore=Chroma.from_documents(
    documents=documents,
    embedding=embedding_model,
    collection_name="Mycollection"
)
```

```python
retriever=vectorStore.as_retriever(search_kwargs={"k":1})
```

```python
query="what is AI ?"
result=retriever.invoke(query)
print(result)
```
