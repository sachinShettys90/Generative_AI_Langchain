```python
from langchain_community.vectorstores import FAISS
from langchain_core.documents import Document
from langchain_openai import OpenAIEmbeddings
```

```python
embedding=OpenAIEmbeddings(api_key="api key")
```

```python
documents=[
    Document(page_content="Langchain is used for LLM"),
    Document(page_content="Langchain is best framework used for AI"),
    Document(page_content="Wikipedia used for searching the data "),
    Document(page_content="AI is very usefull using Chatgpt langchain"),
]
```

```python
vectorStore=FAISS.from_documents(
    documents=documents,
    embedding=embedding
)
```

```python
retriever=vectorStore.as_retriever(
    search_type="mmr",
    search_kwargs={"k":2,"lambda_mult":0.5}
)
```

```python
query="can you tell me about Langchain"
result=retriever.invoke(query)
print(result)
```
